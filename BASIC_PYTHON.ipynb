## Question 1: Reverse a String



```python
# Problem
#Write a Python program to reverse a string.
text = "hello"
reversed_text = text[::-1]
print(reversed_text)
```

```python
##Alternative Method — Using Loop
text='hello'
rev=''
for i in text:
  rev=i+rev
print(rev)
```

##Question 2: Check Palindrome Number
Problem

Write a Python program to check whether a number is a palindrome or not.
Examples:

121 → Palindrome
1331 → Palindrome
123 → Not palindrome

```python
## if it is string
num = input("Enter a number: ")

reverse = ""

for digit in num:
    reverse = digit + reverse

if num == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
## if it is integer
num = int(input("Enter a number: "))

original = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = (reverse * 10) + digit
    num = num // 10

if original == reverse:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

##Question 3: Count Vowels in a String
Problem

Write a Python program to count the number of vowels in a string.

```python
text = input("Enter a string: ")

count = 0

for char in text.lower():
    if char in "aeiou":
        count += 1

print("Number of vowels:", count)
```

## Question 4: Find Largest Number in a List
Problem

Write a Python program to find the largest number in a list.

```python
numbers = [10, 45, 23, 89, 12]

largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

print("Largest number:", largest)
```

Question 5: Find Factorial of a Number

Problem

Write a Python program to find the factorial of a number.

n!=n×(n−1)×(n−2)×⋯×1

Example:

5!=5×4×3×2×1=120

```python
num =5
fac=1
for i in range(1, num+1):
  fac=fac*i
print("Factorial:", fac)

# alternative

import math

num = int(input("Enter a number: "))

print(math.factorial(num))
```

Question 6: Check Prime Number

Problem

Write a Python program to check whether a number is prime or not.

```python
num = int(input("Enter a number: "))

is_prime = True

if num <= 1:
    is_prime = False
else:
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break

if is_prime:
    print("Prime Number")
else:
    print("Not Prime Number")
```

Question 8: Count Frequency of Characters in a String

Problem

Write a Python program to count the frequency of each character in a string.

```python
text = input("Enter a string: ")

frequency = {}

for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

print(frequency)

## Alternative Method Using count()

di={}
for char in set(text):
    di[char] = text.count(char)
print(di)
```
