##Question 2: Check Palindrome Number
Problem

Write a Python program to check whether a number is a palindrome or not.
Examples:

121 → Palindrome
1331 → Palindrome
123 → Not palindrome

```python
## if it is string
num = input("Enter a number: ")

reverse = ""

for digit in num:
    reverse = digit + reverse

if num == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
## if it is integer
num = int(input("Enter a number: "))

original = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = (reverse * 10) + digit
    num = num // 10

if original == reverse:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

##Question 3: Count Vowels in a String
Problem

Write a Python program to count the number of vowels in a string.

```python
text = input("Enter a string: ")

count = 0

for char in text.lower():
    if char in "aeiou":
        count += 1

print("Number of vowels:", count)
```

## 6: Check Prime Number

Problem

Write a Python program to check whether a number is prime or not.

```python
num = int(input("Enter a number: "))

is_prime = True

if num <= 1:
    is_prime = False
else:
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break

if is_prime:
    print("Prime Number")
else:
    print("Not Prime Number")
```

##Question 8: Count Frequency of Characters in a String

Problem

Write a Python program to count the frequency of each character in a string.

```python
text = input("Enter a string: ")

frequency = {}

for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

print(frequency)

## Alternative Method Using count()

di={}
for char in set(text):
    di[char] = text.count(char)
print(di)
```

##Question 9: Find Duplicate Elements in a List

Problem

Write a Python program to find duplicate elements in a list.

```python
numbers = [1, 2, 3, 2, 4, 5, 1, 6]

duplicates = []

for num in numbers:
    if numbers.count(num) > 1 and num not in duplicates:
        duplicates.append(num)

print("Duplicate elements:", duplicates)

## alternative
numbers = [1, 2, 3, 2, 4, 5, 1, 6]

seen = set()
duplicates = set()

for num in numbers:
    if num in seen:
        duplicates.add(num)
    else:
        seen.add(num)

print('duplicate elements',list(duplicates))

```

## Question 10: Sort a List Without Using sort()

Problem

Write a Python program to sort a list in ascending order without using the built-in sort() method.

What is Bubble Sort?

Bubble Sort repeatedly compares adjacent elements and swaps them if they are in wrong order.

Largest elements “bubble up” to the end.

```python
numbers = [5, 2, 8, 1, 9]

n = len(numbers)

for i in range(n):
    for j in range(0, n - i - 1):

        if numbers[j] > numbers[j + 1]:
            numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]

print("Sorted List:", numbers)
```

```python
numbers = [5, 2, 8, 1, 9]

numbers.sort()

print(numbers)
```

##Question 11: Find Second Largest Number in a List
Problem

Write a Python program to find the second largest number in a list.

```python
numbers = [10, 45, 23, 89, 67]

largest = second_largest = float('-inf')

for num in numbers:

    if num > largest:
        second_largest = largest
        largest = num

    elif num > second_largest and num != largest:
        second_largest = num

print("Second Largest:", second_largest)
```

##Question 12: Remove Duplicates from a List

Problem

Write a Python program to remove duplicate elements from a list.

```python
numbers = [1, 2, 3, 2, 4, 1, 5]

unique_list = []

for num in numbers:
    if num not in unique_list:
        unique_list.append(num)

print("List after removing duplicates:", unique_list)
```

##Question 13: Find Sum of Digits of a Number
Problem

Write a Python program to find the sum of digits of a number.

```python
num = int(input("Enter a number: "))

total = 0

while num > 0:
    digit = num % 10
    total += digit
    num = num // 10

print("Sum of digits:", total)
```
